## Nuestra primera visualización 📊

Construyamos una gráfica sencilla con el número de ventas registradas por categoría. Usemos la Libreria MtaplotLIb

```python
ventas_categoria = df["categoria"].value_counts()

ventas_categoria.plot(kind="bar")

plt.title("Número de ventas por categoría")
plt.xlabel("Categoría")
plt.ylabel("Número de ventas")
plt.xticks(rotation=45)
plt.show()
```
